```python
import pandas as pd
import os
from datetime import datetime
```

```python
# Lokasi dataset
file_path = "../data/raw/bali_tourist_destination.csv"

# Membaca dataset CSV
df = pd.read_csv(file_path)

# Tampilkan data
display(df)
```

```python
# Melihat nama seluruh kolom
print("Daftar Kolom:")
print(df.columns.tolist())

# Melihat ukuran dataset
print("Ukuran Dataset:", df.shape)
```

```python
# Menghapus kolom
df_cropped = df.drop(columns=['id_tempat', 'link_gambar'])

# Menampilkan hasil
display(df_cropped)
```

```python
# Path Menyimpan Dataset Process
tanggal_saat_ini = datetime.now()
slug_date = tanggal_saat_ini.strftime("%d-%m-%Y_%H-%M")
output_path = "../data/process/bali_tourist_destination_final_" + slug_date + ".csv"
# print(output_path);

# Simpan data Process
df_cropped.to_csv(output_path, index=False)
print("Dataset berhasil disimpan:", output_path)
```
