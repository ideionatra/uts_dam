```python
import requests
import pandas as pd
import json
import os
from datetime import datetime
```

```python
# URL REST API
url = os.getenv("URL_API")

# Melakukan request GET
response = requests.get(url, timeout=30, params={'limit': 200})

# Memeriksa status request
response.raise_for_status()

# Memetakan response menjadi JSON dan print
result = response.json()
print(json.dumps(result, indent=4, ensure_ascii=False))
```

```python
# Mengambil seluruh data pada respons
data = result["data"]

# Mengubah JSON menjadi DataFrame
df = pd.DataFrame(data)
display(df)

# Simpan Data Raw
tanggal_saat_ini = datetime.now()
slug_date = tanggal_saat_ini.strftime("%d-%m-%Y_%H-%M")
output_path = "../data/raw/wisata_bahari_indonesia_" + slug_date + ".csv"
print(output_path);

# Simpan data Raw
df.to_csv(output_path, index=False)
print("Dataset Raw berhasil disimpan:", output_path)
```

```python
# Melihat daftar kolom
print("\nDaftar Kolom:")
print(df.columns.tolist())

# Melihat jumlah baris dan kolom
print("Jumlah data:", len(df))
print("Shape:", df.shape)

# Membandingkan dengan total dari API
print("Total data API:", result["total_count"])
```

```python
# Kolom yang akan dihapus
kolom_hapus = [
    "created_on",
    "created_by",
    "updated_on",
    "updated_by"
]

# Menghapus kolom
df_cropped = df.drop(columns=kolom_hapus, errors="ignore")
df_order_by_year_old = df_cropped.sort_values(by="tahun", ascending=True)


# Menampilkan hasil preprocessing
display(df_cropped)
display(df_order_by_year_old)

# Memeriksa shape terbaru
print("Shape setelah preprocessing:", df_cropped.shape)
```

```python
# Simpan Data Process
tanggal_saat_ini = datetime.now()
slug_date = tanggal_saat_ini.strftime("%d-%m-%Y_%H-%M")
output_path_cropped = "../data/process/wisata_bahari_indonesia_" + slug_date + ".csv"
output_path_order_by_year = "../data/process/wisata_bahari_indonesia_order-by-year_" + slug_date + ".csv"
# print(output_path);

# Simpan data Process
df_cropped.to_csv(output_path_cropped, index=False)
df_order_by_year_old.to_csv(output_path_order_by_year, index=False)
print("Dataset process berhasil disimpan:", output_path)
```
